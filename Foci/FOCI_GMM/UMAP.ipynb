### PERANOVA and PERMDISP for all data

```python
# Groupping variable for PERMANOVA and PERMDISP
variable = "sample"

# Calculate distances
X = df[features].copy()
X_scaled = StandardScaler().fit_transform(X)
    
distances = squareform(
    pdist(X_scaled, metric="euclidean")
)

dm = DistanceMatrix(
    distances,
    ids=[str(i) for i in range(len(df))]
)

# PERMANOVA
perm = permanova(
    dm,
    grouping=df[variable].to_numpy(),
    permutations=9999,
    seed=42
)

# PERMDISP
disp = permdisp(
    dm,
    grouping=df[variable].to_numpy(),
    permutations=9999,
    seed=42
)

# Calculate R²
F = perm["test statistic"]
n = len(df)
k = df[variable].nunique()

R2 = (F * (k - 1)) / (
    F * (k - 1) + (n - k)
)

print("PERMANOVA")
print(f"F = {F:.4f}")
print(f"p = {perm['p-value']:.4f}")
print(f"R² = {R2:.4f}")

print("\nPERMDISP")
print(f"F = {disp['test statistic']:.4f}")
print(f"p = {disp['p-value']:.4f}")
```

## PERMANOVA for all clusters together

```python
# Select features and groups
data = umap_df[features + ["sample"]].dropna().copy()

X = data[features]
groups = data["sample"].to_numpy()

# Standardize features
X_scaled = StandardScaler().fit_transform(X)

# Euclidean distance matrix
distances = squareform(
    pdist(X_scaled, metric="euclidean")
)

dm = DistanceMatrix(distances)

# PERMANOVA
permanova_result = permanova(
    dm,
    grouping=groups,
    permutations=9999,
    seed=42
)

# PERMDISP
permdisp_result = permdisp(
    dm,
    grouping=groups,
    test="centroid",
    permutations=9999,
    seed=42
)

# Calculate R²
F = permanova_result["test statistic"]
n = len(groups)
k = len(np.unique(groups))

R2 = (F * (k - 1)) / (F * (k - 1) + (n - k))

# Results dataframe
results_df = pd.DataFrame({
    "Test": ["PERMANOVA", "PERMDISP"],
    "Statistic": [
        permanova_result["test statistic"],
        permdisp_result["test statistic"]
    ],
    "p_value": [
        permanova_result["p-value"],
        permdisp_result["p-value"]
    ],
    "R2": [R2, np.nan],
    "n": [n, n],
    "groups": [k, k]
})

display(results_df.round(4))
```

# Xi-statistics

```python
from scipy.stats import chi2_contingency, fisher_exact
from statsmodels.stats.multitest import multipletests
import pandas as pd

# counts is your existing contingency table
display(counts)

# Overall chi-square test
chi2, p, dof, expected = chi2_contingency(counts)

print(f"Chi-square = {chi2:.4f}")
print(f"p-value = {p:.6f}")

# Pairwise comparisons
comparisons = [
    ("WT", "MGS3"),
    ("WT", "WT, NC"),
    ("MGS3", "MGS3, NC"),
    ("WT, NC", "WT, Apt.3"),
    ("MGS3, NC", "MGS3, Apt.3"),
    ("WT, NC", "MGS3, NC"),
    ("WT, Apt.3", "MGS3, Apt.3")
]

results = []

for group1, group2 in comparisons:

    contingency = counts.loc[
        [group1, group2], [0, 1]
    ]

    odds_ratio, p_value = fisher_exact(contingency)

    results.append({
        "comparison": f"{group1} vs {group2}",
        "odds_ratio": odds_ratio,
        "p_value": p_value
    })

results_df = pd.DataFrame(results)

results_df["p_adjusted"] = multipletests(
    results_df["p_value"],
    method="fdr_bh" # Benjamini–Hochberg
)[1]

display(results_df)
```

### Centroid distance visualization for each cluster separately

```python
# Settings
reference = "WT"
group_col = "sample"
cluster_col = "cluster"

# Use the same features as in your PERMANOVA
# features = [...]

# Keep only valid rows
df = umap_df.dropna(
    subset=features + [group_col, cluster_col]
).copy()

# Standardize features globally
# This makes distances comparable between clusters
scaler = StandardScaler()
df[features] = scaler.fit_transform(df[features])

# Consistent colors for mutants
groups = sorted(df[group_col].unique())

palette = dict(zip(
    groups,
    sns.color_palette("Set2", len(groups))
))

# Calculate centroid distances
distance_results = []

for cluster, subset in df.groupby(cluster_col):

    if reference not in subset[group_col].values:
        continue

    # Centroid of each sample
    centroids = subset.groupby(group_col)[features].mean()

    # WT centroid
    ref_centroid = centroids.loc[reference].values

    for group in centroids.index:

        if group == reference:
            continue

        distance = np.linalg.norm(
            centroids.loc[group].values - ref_centroid
        )

        distance_results.append({
            "cluster": cluster,
            "sample": group,
            "distance": distance,
            "n": (subset[group_col] == group).sum()
        })

distance_df = pd.DataFrame(distance_results)

print(distance_df)

# Plot centroid distance trees
clusters = sorted(distance_df["cluster"].unique())

ncols = min(2, len(clusters))
nrows = math.ceil(len(clusters) / ncols)

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(9 * ncols, 6 * nrows),
    squeeze=False
)

axes = axes.flatten()

max_distance = distance_df["distance"].max()

for ax, cluster in zip(axes, clusters):

    data = distance_df[
        distance_df["cluster"] == cluster
    ].sort_values("distance")

    y_positions = np.arange(len(data))

    for y, (_, row) in zip(y_positions, data.iterrows()):

        group = row["sample"]
        distance = row["distance"]

        # Line from WT to mutant centroid
        ax.plot(
            [0, distance],
            [y, y],
            color=palette[group],
            linewidth=2.5,
            alpha=0.8
        )

        # WT reference
        ax.scatter(
            0, y,
            color="black",
            s=70,
            zorder=3
        )

        # Mutant centroid
        ax.scatter(
            distance, y,
            color=palette[group],
            s=160,
            edgecolor="black",
            linewidth=0.8,
            zorder=4
        )

        # Distance label
        ax.text(
            distance + max_distance * 0.02,
            y,
            f"{distance:.2f}",
            va="center",
            fontsize=12
        )

    ax.set_yticks(y_positions)
    ax.set_yticklabels(
        data["sample"],
        fontsize=13
    )

    ax.set_xlim(
        -max_distance * 0.05,
        max_distance * 1.20
    )

    ax.set_xlabel(
        "Euclidean distance from WT centroid",
        fontsize=13
    )

    ax.set_title(
        f"Cluster {cluster}",
        fontsize=17,
        fontweight="bold"
    )

    ax.grid(
        axis="x",
        alpha=0.2
    )

    ax.spines[["top", "right"]].set_visible(False)

# Remove empty subplots
for ax in axes[len(clusters):]:
    ax.remove()

plt.tight_layout(pad=3)
plt.show()
```

## Difference between clusters

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
from sklearn.preprocessing import StandardScaler


features = [
    "n_foci",
    "area",
    "equivalent_diameter_area",
    "eccentricity",
    "perimeter",
    "solidity",
    "mean_intensity",
    "max_intensity",
    "min_intensity",
]


# Keep only the two clusters
clusters = sorted(umap_df["cluster"].dropna().unique())

assert len(clusters) == 2, "Expected exactly two clusters"

df_analysis = umap_df.dropna(
    subset=["cluster"] + features
).copy()

# Summary statistics
summary = df_analysis.groupby("cluster")[features].agg(
    ["median", "std", "count"]
)

#display(summary.round(3))

results = []

c0, c1 = clusters

for feature in features:

    x = df_analysis.loc[
        df_analysis["cluster"] == c0, feature
    ].dropna()

    y = df_analysis.loc[
        df_analysis["cluster"] == c1, feature
    ].dropna()

    # Cohen's d
    pooled_sd = np.sqrt(
        ((len(x)-1)*x.var(ddof=1) +
         (len(y)-1)*y.var(ddof=1)) /
        (len(x)+len(y)-2)
    )

    d = (
        (y.median() - x.median()) / pooled_sd
        if pooled_sd > 0 else np.nan
    )

    # Mann–Whitney U
    stat, p = mannwhitneyu(
        x, y,
        alternative="two-sided"
    )

    results.append({
        "feature": feature,
        f"mean_cluster_{c0}": x.mean(),
        f"mean_cluster_{c1}": y.mean(),
        "cohens_d": d,
        "abs_effect_size": abs(d),
        "p_value": p
    })

results_df = pd.DataFrame(results)

# Multiple-testing correction
results_df["p_adjusted"] = multipletests(
    results_df["p_value"],
    method="fdr_bh"
)[1]

results_df = results_df.sort_values(
    "abs_effect_size",
    ascending=False
)

display(results_df.round(2))
```
