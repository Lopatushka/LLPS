```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
def scatterplot(x, y, xlabel, ylabel):
    plt.scatter(x, y, marker="o")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()
```

```python
DIR = "/mnt/c/users/helen/Desktop/test"
FILE_NAME = "C1_WT-02-Airyscan_Processing-30_ROI_0649-0821_foci.csv"
```

```python
# Search recursively
paths = list(Path(DIR).rglob(FILE_NAME))

print(f"Found {len(paths)} files:")
for path in paths:
    print(path)

# Read every file as a DataFrame
dfs = [pd.read_csv(path) for path in paths]
```

```python
X = [1, 10, 25, 5, 50]
```

```python
n_foci = [len(df) for df in dfs]

scatterplot(x=X, y=n_foci, xlabel="x SD", ylabel="Number of foci")
```

```python
sigma_mean = [df['sigma [nm]'].mean() for df in dfs]

scatterplot(x=X, y=sigma_mean, xlabel="x SD", ylabel="Mean Sigma [nm]")
```

```python
photons = [df['intensity [photon]'].mean() for df in dfs]

scatterplot(x=X, y=photons, xlabel="x SD", ylabel="Mean Photon intensity [photons]")
```
